```r
library(data.table)

# 1. Crear el Dataset de Experimentos (Ejemplo estructurado)
# Reemplazar con los vectores reales de resultados por semilla
resultados_semillas <- data.table(
  semilla      = c(100019, 200003, 300007, 400031, 500069, 600097, 700101, 800107, 900121, 100151),
  auc_baseline = c(0.9228, 0.9215, 0.9230, 0.9210, 0.9241, 0.9220, 0.9218, 0.9235, 0.9222, 0.9229),
  auc_con_ga   = c(0.9245, 0.9238, 0.9231, 0.9225, 0.9250, 0.9228, 0.9230, 0.9234, 0.9239, 0.9242)
)

# 2. Aplicar el Test de Wilcoxon Pareado
# alternative = "greater" evalúa si 'auc_con_ga' es significativamente MAYOR que 'auc_baseline'
test_res <- wilcox.test(
  x = resultados_semillas$auc_con_ga,
  y = resultados_semillas$auc_baseline,
  paired = TRUE,
  alternative = "greater",
  exact = FALSE # Usar FALSE si hay empates (ties) entre diferencias
)

# 3. Presentación de Resultados
cat("=== Test de Wilcoxon Pareado ===\n")
cat("Estadístico V:", test_res$statistic, "\n")
cat("p-value:", test_res$p.value, "\n\n")

if (test_res$p.value < 0.05) {
  cat("Resultado: La mejora es ESTADÍSTICAMENTE SIGNIFICATIVA (p < 0.05).\n")
} else {
  cat("Resultado: NO hay evidencia suficiente de mejora significativa (p >= 0.05).\n")
}
```
